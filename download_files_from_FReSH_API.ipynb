# Get studies Ids list

```python
import requests 
import json

# Get all studies from FRESH API and save to file
url_api = "https://nada.portail-fresh.fr/index.php/api/catalog/search?ps=10000"
response = requests.get(url_api)
json_content = response.json()

# write json content to file
with open('data/FRESH-API-all_studies.json', 'w') as f:
    json.dump(json_content, f, indent=4)

```

```python
all_studies_dict = json.loads(json.dumps(json_content))
studies_list = all_studies_dict['result']["rows"]

# loop through studies and store ids in a list
ids_list = []
for study in studies_list:
    ids_list.append(study['idno'])

print(f"List of study IDs: {ids_list}")
```

## get all studies xml

```python
# loop through all study ids and get xml content for each study
HEADERS_XML = {
    "Accept": "application/xml"
}

for id in ids_list:
    url_api_study = f"https://nada.portail-fresh.fr/index.php/api/catalog/{id}"
    response_study = requests.get(url_api_study, headers=HEADERS_XML, timeout=30)
    filename = f'data/input/{id}.xml'
    with open(filename, 'wb') as f:
        f.write(response_study.content)
    print(f"Downloaded XML for study ID: {id} to file: {filename}")
```
